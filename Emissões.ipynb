```python
df = pd.read_csv('emissions.csv', sep=' ', header=None, names=['country', 'year', 'emissions'])
df.head()
```

```python
china = df[(df['country'] == "CHN") & (df['year'] > 1950)]
brasil = df[(df['country'] == "BRA") & (df['year'] > 1950)]
usa = df[(df['country'] == "USA") & (df['year'] > 1950)]
india = df[(df['country'] == "IND") & (df['year'] > 1950)]
```

```python
china.emissions.sum()
```

```python
brasil.emissions.sum()
```

```python
plt.plot(brasil['year'], brasil['emissions'], label='Brasil')
plt.plot(usa['year'], usa['emissions'], label='USA')
plt.plot(china['year'], china['emissions'], label='China')
plt.plot(india['year'], india['emissions'], label='India')
plt.title('Emissões do Brasil após 2000')
plt.xlabel('Ano')
plt.ylabel('Emissões')
plt.legend()
plt.show()
```
